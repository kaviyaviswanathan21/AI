# How many mushrooms live in a numerous population?

To solve this, I will find the total number of mushrooms and find the ones among them that live in a numerous population. This question can be answered with **frequency** because you would just have to count the number of mushrooms that live in a numerous population.


```python
import pandas as pd

fungi = pd.read_csv("mushrooms.csv")
fungi.head()
```

```python
import pandas as pd

fungi = pd.read_csv("mushrooms.csv")

fungi["population"].value_counts()
```

```python
print("The total number of Mushrooms that live in a numerous population: 400")
```

# Visual:

```python
import matplotlib.pyplot as plt
labels = 'v', 'y', 's', 'n','a','c'
sizes= [4040,1712, 1248,400,384,340]
fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%',
       pctdistance=1.25, labeldistance=.6)
ax.pie(sizes, labels=labels,
       colors=['pink', 'lightblue', 'purple', 'red', 'blue', 'yellow'])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

population_counts = fungi["population"].value_counts().reset_index()
population_counts.columns = ["population", "count"]

sns.barplot(data=population_counts, x="population", y="count")

plt.xlabel("Population")
plt.ylabel("Number of Mushrooms")
plt.title("Number of Mushrooms by Population")
plt.show()
```
