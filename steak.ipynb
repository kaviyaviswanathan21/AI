# Does Gender have an Affect on how they Like Their Steak Cooked?

To solve this, I will be conducting an **observational study** by looking at two variables on the table that show the gender and how they like their steak prepared and see if your gender plays a role in the type of steak you prefer to eat. 

```python
import pandas as pd

steak = pd.read_csv("steak.csv")
steak.head()
```

### Overview:

```python
steak.info()
```

### My Study:

```python
pd.crosstab(steak["Gender"],steak["How do you like your steak prepared?"])
```

### Visual:

```python
pd.crosstab(steak["Gender"], steak["How do you like your steak prepared?"]).plot(kind="bar")
```

This bar graph gives a visual on the prefered steak for males and females. Looking at the bars, there is not a significant difference on gender and how they like their steak.

# Conclusion:
Based on my study, I don't think there is a significant connection between your gender and how you like your steak cooked because the results for the type of steak and your gender were fairly similar to each other. For example, the frequency for females who prefer medium rare was 77 and the frequency for males who prefer medium rare was 81, showing that medium rare is the most prefered steak among both genders.
